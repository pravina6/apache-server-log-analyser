```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read CSV files

ip_df = pd.read_csv("ip_frequency.csv")
status_df = pd.read_csv("http_status_frequency.csv")
```

```python
# Plot top 10 IP addresses


ip_df = ip_df.sort_values("Frequency", ascending=False).head(10)
plt.figure(figsize=(10, 5))
plt.bar(ip_df["Value"], ip_df["Frequency"],color="grey")
plt.title("Top 10 IP Addresses by Frequency")
plt.xlabel("IP Address")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig("ip_frequency.png")
plt.show()
```

```python
# Plot HTTP status code distribution


status_df = status_df.sort_values("Value")
plt.figure(figsize=(8, 5))
plt.bar(status_df["Value"], status_df["Frequency"])
plt.title("HTTP Status Code Distribution")
plt.xlabel("HTTP Status Code")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("status_frequency.png")
plt.show()

print("Charts saved successfully!")
```
