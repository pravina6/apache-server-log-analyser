```python
import pandas as pd
import matplotlib.pyplot as plt
```

##### read csv files

```python
ip_df = pd.read_csv("ip_frequency.csv")
status_df=pd.read_csv("http_status_frequency.csv")
```

##### Find most common IP address

```python
ip_df = ip_df.sort_values("Frequency")#, ascending=False)
```

```python
print("Most Common IP Address:", ip_df.iloc[0]["Value"])
print("Frequency:", ip_df.iloc[0]["Frequency"])
```

#####  Find most common HTTP status code

```python
#status_df = status_df.sort_values("Frequency", ascending=False)

print("\nHTTP Status Code Frequencies:")
print(status_df.sort_values("Frequency"))#ascending=False))
```

```python
print("\nMost Common HTTP Status Code:", status_df.iloc[0]["Value"])
print("Frequency:", status_df.iloc[0]["Frequency"])
```

##### Visualize top 10 IP addresses

```python
top_ip = ip_df.head(10)
```

```python
plt.figure(figsize=(12, 6))
plt.bar(top_ip["Value"], top_ip["Frequency"], color="grey")
plt.title("Top 10 Most Common IP Addresses")
plt.xlabel("IP Address")
plt.ylabel("Frequency")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

#####  Visualize HTTP status code frequency

```python
plt.figure(figsize=(12, 6))
plt.bar(status_df["Value"].astype(str), status_df["Frequency"], color="crimson")
plt.title("HTTP Status Code Frequency")
plt.xlabel("HTTP Status Code")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
